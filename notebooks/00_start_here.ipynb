## Step 1 — Prepare a reproducible experiment

A seed makes comparisons easier. Exact numerical results can still differ across devices and PyTorch versions.


```python
import random
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
import torch
from torch import nn
from torch.utils.data import DataLoader, TensorDataset

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.set_num_threads(2)  # Small examples often run faster with fewer CPU threads.
device = torch.device("cpu")  # Learn on CPU first; the performance lab covers CUDA.
plt.rcParams.update({"figure.figsize": (7, 4), "axes.grid": True})
print("PyTorch:", torch.__version__, "| Device:", device)
```

## Step 3 — Turn numbers into a tensor

A tensor is a collection of numbers with a shape. Here each row is one observation. We use floating point numbers because learning updates usually involve fractions.


```python
hours = torch.tensor([[1.0], [2.0], [3.0], [4.0]])
scores = 3 * hours + 2
print("Input shape:", hours.shape, "dtype:", hours.dtype)
print("Targets:", scores.squeeze().tolist())
assert hours.shape == (4, 1)
```

## Step 4 — Make a prediction and measure its error

Our first model is a rule: prediction = weight × hours + bias. **Parameters** are the numbers we can adjust. **Loss** is one number summarizing the error; smaller is better for this metric.


```python
weight, bias = 1.0, 0.0
predictions = weight * hours + bias
loss = ((predictions - scores) ** 2).mean()
print("Predictions:", predictions.squeeze().tolist())
print("Mean squared error:", loss.item())
plt.scatter(hours[:, 0], scores[:, 0], label="target")
plt.plot(hours[:, 0], predictions[:, 0], label="our rule", color="orange")
plt.xlabel("Hours (toy example)"); plt.ylabel("Score"); plt.legend(); plt.show()
```

## Step 5 — Improve the rule manually

Learning means adjusting parameters to reduce loss. Here you already know the generating rule, so you can do it manually. Later, autograd and an optimizer will do the arithmetic.


```python
better_predictions = 3.0 * hours + 2.0
better_loss = ((better_predictions - scores) ** 2).mean()
print("Before:", loss.item(), "After:", better_loss.item())
assert better_loss < loss
```
