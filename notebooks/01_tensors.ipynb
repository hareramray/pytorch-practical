## Step 1 — Prepare a reproducible experiment

A seed makes comparisons easier. Exact numerical results can still differ across devices and PyTorch versions.


```python
import random
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
import torch
from torch import nn
from torch.utils.data import DataLoader, TensorDataset

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.set_num_threads(2)  # Small examples often run faster with fewer CPU threads.
device = torch.device("cpu")  # Learn on CPU first; the performance lab covers CUDA.
plt.rcParams.update({"figure.figsize": (7, 4), "axes.grid": True})
print("PyTorch:", torch.__version__, "| Device:", device)
```

## Step 2 — Create tensors and inspect them

A scalar has zero dimensions, a vector has one, and a matrix has two. `shape` describes the axes; `dtype` describes the stored numbers; `device` describes where they live.


```python
scalar = torch.tensor(7.)
vector = torch.arange(5, dtype=torch.float32)
matrix = torch.arange(12, dtype=torch.float32).reshape(3, 4)
for name, t in [("scalar", scalar), ("vector", vector), ("matrix", matrix)]:
    print(name, "shape:", tuple(t.shape), "dtype:", t.dtype, "device:", t.device)
print(torch.zeros(2, 3))
print(torch.ones_like(matrix))
```

## Step 3 — Select, mask, and reduce

Indexing selects values; reductions summarize them. For a matrix `[rows, columns]`, `mean(dim=0)` collapses the row axis and leaves one mean for each column.


```python
print("First row:", matrix[0])
print("Second column:", matrix[:, 1])
print("Last two rows:", matrix[-2:])
print("Values above 5:", matrix[matrix > 5])
print("Column means:", matrix.mean(dim=0))
print("Row means:", matrix.mean(dim=1))
assert matrix.mean(dim=0).shape == (4,)
```

## Step 4 — Distinguish a view from a copy

Some operations share storage. Changing a view changes the original. Use `clone()` when you need independent data. `torch.from_numpy` also shares CPU memory with its NumPy array.


```python
original = torch.tensor([1., 2., 3.])
view = original[:2]
independent = original.clone()
view[0] = 99
print("Original:", original, "Copy:", independent)
array = np.array([1., 2.], dtype=np.float32)
shared = torch.from_numpy(array)
array[0] = 8
assert shared[0].item() == 8
assert independent[0].item() == 1
```

## Step 5 — Move data deliberately

Operations generally require tensors on the same device. `.to(...)` returns the converted tensor, so keep the result. Class indices usually use `long`; learned numeric features usually use floating point.


```python
labels = torch.tensor([0, 1, 2], dtype=torch.long)
features = matrix.to(device=device, dtype=torch.float32)
numpy_copy = features.detach().cpu().numpy().copy()
print("Features:", features.dtype, "Labels:", labels.dtype)
assert np.allclose(numpy_copy, matrix.numpy())
```
